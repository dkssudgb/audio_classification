#### Metadata 가져오기

```python
# 경로 : dir1 > 라벨링데이터 > 파일명
# json 파일 미리보기
# "LabelDataInfo" - "Path" : 경로, "Desc" : 음성 종류
with open(path + dir1[0] + "/" + "라벨링데이터/S-210915_S_101_D_002_0001.json") as f:
    data = json.load(f)
    print(data)
```

```python
# json 파일에서 Path와 Desc만 불러와서 데이터프레임으로

def file_to_df(path):
    fname = glob(path)
    file_name = []
    data_class = []
    for file in fname:
        with open(file, 'r') as f:
            data = json.load(f)
            file_name.append(data['LabelDataInfo']['LabelID'] + '.mp3')
            data_class.append(data['LabelDataInfo']['Desc'])
    df = pd.DataFrame({'file_name': file_name, 'class': data_class})
    return df
```

```python
train_meta = file_to_df(path + dir1[0] + "/라벨링데이터/*")
valid_meta = file_to_df(path + dir1[1] + "/라벨링데이터/*")

aihub = pd.concat([train_meta, valid_meta])
print(aihub.shape)
aihub.head()
```

```python
# class 값 수정
file_names.replace({'car_horn':'자동차 경적'}, inplace=True)
file_names.replace({'자동차 경적 소리':'자동차 경적'}, inplace=True)

file_names.replace({'dog_bark':'개 짖는 소리'}, inplace=True)

file_names.replace({'경찰차 사이렌 소리':'경찰차 사이렌'}, inplace=True)

file_names.replace({'소방차 사이렌 소리':'소방차 사이렌'}, inplace=True)
file_names.replace({'전주 완산 소방서 화제 사이렌 소리':'소방차 사이렌'}, inplace=True)

file_names.replace({'응급경보 소리':'응급 경보'}, inplace=True)
file_names.replace({'응급경보 사이렌 소리':'응급 경보'}, inplace=True)

file_names.replace({'구급차 사이렌 소리':'구급차 사이렌'}, inplace=True)

file_names.replace({'화재경보 소리':'화재 경보'}, inplace=True)
file_names.replace({'화재 경보 소리':'화재 경보'}, inplace=True)

file_names.replace({'도난경보 소리':'도난 경보'}, inplace=True)
file_names.replace({'도난 경보음 소리':'도난 경보'}, inplace=True)

file_names.replace({'가스누설 화재경보 소리':'가스누설 화재 경보'}, inplace=True)

file_names.replace({'침입감지 경보 소리':'침입 경보'}, inplace=True)

file_names.replace({'공습경보 소리':'공습 경보'}, inplace=True)
```

```python
file_names['class'].value_counts()
```

```python
# 위험소리가 아니거나 데이터 수가 1개인 경우 제거
drop_class = ['비상경보 소리', '철도 건널목 신호음 소리', '렉카차 사이렌 소리', '민방위훈련 사이렌 소리', '소방서 구급 경보 사이렌 소리', '소방서 경찰 순찰차 사이렌 소리', '재난경보 소리', '해안 경찰 보트 사이렌 소리']
file_names = file_names[~file_names['class'].isin(drop_class)]

file_names['class'].value_counts()
```

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

file_names['class_num'] = encoder.fit_transform(file_names['class'])
file_names.head()
```
